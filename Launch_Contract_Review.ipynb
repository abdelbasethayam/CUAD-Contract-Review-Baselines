# Launch Contract Review Demo

Run the next cell once. It starts the backend and frontend if needed, checks both health endpoints, and prints the link to open the app.

```python
from pathlib import Path
import subprocess
repo = Path.cwd()
while repo != repo.parent and not (repo / "launch_contract_review.sh").exists():
    repo = repo.parent
if not (repo / "launch_contract_review.sh").exists():
    raise FileNotFoundError("Open this notebook from the CUAD-Contract-Review-Baselines repository, or set repo to its folder.")
result = subprocess.run(["bash", str(repo / "launch_contract_review.sh")], cwd=repo, text=True, capture_output=True)
print(result.stdout)
if result.stderr:
    print(result.stderr)
result.check_returncode()
from IPython.display import display, HTML
display(HTML('<a href="http://127.0.0.1:5173/" target="_blank" style="display:inline-block;padding:12px 18px;background:#1769aa;color:white;border-radius:8px;text-decoration:none;font-weight:600">Open Contract Review App</a>'))
```
