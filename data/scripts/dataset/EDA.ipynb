```python
# Categories suitable for clause classification
CLAUSE_CATEGORIES = [
    "Governing Law",
    "Most Favored Nation",
    "Non-Compete",
    "Exclusivity",
    "No-Solicit of Customers",
    "Competitive Restriction Exception",
    "No-Solicit of Employees",
    "Non-Disparagement",
    "Termination for Convenience",
    "Rofr/Rofo/Rofn",
    "Change of Control",
    "Anti-Assignment",
    "Revenue/Profit Sharing",
    "Price Restrictions",
    "Minimum Commitment",
    "Volume Restriction",
    "IP Ownership Assignment",
    "Joint IP Ownership",
    "License Grant",
    "Non-Transferable License",
    "Affiliate License-Licensor",
    "Affiliate License-Licensee",
    "Unlimited/All-You-Can-Eat-License",
    "Irrevocable or Perpetual License",
    "Source Code Escrow",
    "Post-Termination Services",
    "Audit Rights",
    "Uncapped Liability",
    "Cap on Liability",
    "Liquidated Damages",
    "Insurance",
    "Covenant Not to Sue",
    "Third Party Beneficiary",
]


# Categories that are mainly information extraction tasks
EXTRACTION_CATEGORIES = [
    "Document Name",
    "Parties",
    "Agreement Date",
    "Effective Date",
    "Expiration Date",
    "Renewal Term",
    "Notice Period To Terminate Renewal",
    "Warranty Duration",
]
```

```python
def find_column(category, suffix=""):
    """
    Find the actual CUAD CSV column corresponding to a category.
    Handles small naming differences between CUAD versions.
    """

    target = f"{category}{suffix}".strip().lower()

    # Exact match first
    for col in df.columns:
        if str(col).strip().lower() == target:
            return col

    # Normalized comparison
    def normalize(text):
        return (
            str(text)
            .lower()
            .replace("-", "")
            .replace("_", "")
            .replace(" ", "")
        )

    normalized_target = normalize(target)

    for col in df.columns:
        if normalize(col) == normalized_target:
            return col

    return None
```

```python
print("\n" + "=" * 70)
print("CATEGORY VERIFICATION")
print("=" * 70)

print("\nClause categories:")

missing_clause_categories = []

for category in CLAUSE_CATEGORIES:
    context_col = find_column(category)

    if context_col:
        print(f"[OK] {category:<45} -> {context_col}")
    else:
        print(f"[MISSING] {category}")
        missing_clause_categories.append(category)


print("\nExtraction categories:")

missing_extraction_categories = []

for category in EXTRACTION_CATEGORIES:
    context_col = find_column(category)
    answer_col = find_column(category, "-Answer")

    if context_col and answer_col:
        print(
            f"[OK] {category:<45} -> "
            f"{context_col} | {answer_col}"
        )
    else:
        print(
            f"[MISSING] {category} | "
            f"context={context_col} answer={answer_col}"
        )
        missing_extraction_categories.append(category)
```

```python
print("\n" + "=" * 70)
print("BUILDING CLAUSE CLASSIFICATION DATASET")
print("=" * 70)

classification_rows = []

for _, row in df.iterrows():

    document_id = row["Filename"]

    for category in CLAUSE_CATEGORIES:

        context_col = find_column(category)

        if context_col is None:
            continue

        clause_text = row[context_col]

        # Ignore empty / missing clauses
        if pd.isna(clause_text):
            continue

        clause_text = str(clause_text).strip()

        if not clause_text:
            continue

        # Ignore obvious missing values
        if clause_text.lower() in {"nan", "none", "null"}:
            continue

        answer_col = find_column(category, "-Answer")

        answer = None

        if answer_col:
            value = row[answer_col]

            if pd.notna(value):
                answer = str(value).strip()

        classification_rows.append({
            "document_id": document_id,
            "clause_type": category,
            "clause_text": clause_text,
            "answer": answer,
        })


classification_df = pd.DataFrame(classification_rows)
```

```python
print("\n" + "=" * 70)
print("BUILDING INFORMATION EXTRACTION DATASET")
print("=" * 70)

extraction_rows = []

for _, row in df.iterrows():

    document_id = row["Filename"]

    for category in EXTRACTION_CATEGORIES:

        context_col = find_column(category)
        answer_col = find_column(category, "-Answer")

        if context_col is None or answer_col is None:
            continue

        context = row[context_col]
        answer = row[answer_col]

        # We need an actual labeled example
        if pd.isna(context) and pd.isna(answer):
            continue

        context = "" if pd.isna(context) else str(context).strip()
        answer = "" if pd.isna(answer) else str(answer).strip()

        # Ignore completely empty records
        if not context and not answer:
            continue

        extraction_rows.append({
            "document_id": document_id,
            "category": category,
            "context": context,
            "answer": answer,
        })


extraction_df = pd.DataFrame(extraction_rows)
```

```python
from pathlib import Path

# Output directory
output_dir = Path("dataset/processed")
output_dir.mkdir(parents=True, exist_ok=True)

# Output paths
classification_path = output_dir / "clause_classification.csv"
extraction_path = output_dir / "information_extraction.csv"

# Save classification dataset
classification_df.to_csv(
    classification_path,
    index=False,
    encoding="utf-8-sig"
)

# Save extraction dataset
extraction_df.to_csv(
    extraction_path,
    index=False,
    encoding="utf-8-sig"
)

print("Saved successfully!")
print(f"Classification: {classification_path}")
print(f"Extraction:     {extraction_path}")
```
